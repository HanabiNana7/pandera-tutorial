```python
import pandas as pd
import pandera.pandas as pa
from pandera.pandas import Column, DataFrameSchema, Check
```

# DataFrameSchema

## 定義 Schema（資料驗證規則）

```python
def validate_dataframe(
    df_to_validate: pd.DataFrame,
    schema: pa.DataFrameSchema,
) -> None:
    """
    Validates the input DataFrame against the given schema.
    驗證輸入的 DataFrame 是否符合給定的 schema

    Args:
        df_to_validate (pd.DataFrame): The DataFrame to be validated.
        schema (pa.DataFrameSchema): The Pandera schema to validate the DataFrame.
    """
    try:
        validated_df = schema.validate(df_to_validate)
        print("✅ Validation successful. DataFrame:")
        print(validated_df)
    except (pa.errors.SchemaError, pa.errors.SchemaErrors) as e:
        print("❌ Validation failed:")
        if hasattr(e, "failure_cases"):
            print(e.failure_cases)
        else:
            print(e)
```

```python
schema = DataFrameSchema(
    {
        "id": Column(int, Check.greater_than(0)),
        "name": Column(str, Check.str_length(min_value=1)),
        "age": Column(int, Check.in_range(18, 99), nullable=False),
        "email": Column(str, Check.str_matches(r".+@.+\..+"), nullable=True),
    }
)
```

## 檢查資料

### 正確資料

```python
correct_df = pd.DataFrame(
    {
        "id": [1, 2],
        "name": ["MilkTea", "BlackTea"],
        "age": [25, 30],
        "email": ["milktea@example.com", "blacktea@example.com"],
    }
)

validate_dataframe(df_to_validate=correct_df, schema=schema)
```

### 錯誤資料

```python
correct_df = pd.DataFrame(
    {
        "id": [1, 2],
        "name": ["MilkTea", "BlackTea"],
        "age": [25, 30],
        "email": ["milktea", "blacktea"],
    }
)

validate_dataframe(df_to_validate=correct_df, schema=schema)
```
